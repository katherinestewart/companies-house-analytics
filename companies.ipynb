### Check Company Data SIC Codes are all 5 digits

```python
all_sics = df[sic_cols].stack().dropna()

sic_parts = all_sics.str.split(" - ", n=1, expand=True)
sic_parts.columns = ["SICCode", "Description"]

sic_parts["SICCode"] = sic_parts["SICCode"].str.strip()
sic_parts["Description"] = sic_parts["Description"].str.strip()

sic_parts.head()
```

```python
sic_parts["SICCode"].str.len().value_counts()
```

```python
sic_parts["SICCode"].str.isdigit().value_counts()
```

### Question 2:

**Why do we have 2981 four-digit numeric codes**

```python
(
    sic_parts.loc[
        sic_parts["SICCode"].str.len() == 4,
        ["SICCode", "Description"]
    ]
    .drop_duplicates()
    .sort_values("SICCode")
)
```
